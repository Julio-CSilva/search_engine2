```python
import xml.etree.ElementTree as ET

def extract_tags(xml_file):
    tags = set()
    values = set()
    
    try:
        tree = ET.parse(xml_file)
        root = tree.getroot()
        
        for element in root.iter():
            tags.add(element.tag)
            values.add(element.text.strip() if element.text else "")
        
        return tags, values
    except Exception as e:
        print(f"Erro ao processar o arquivo: {e}")
        return []

if __name__ == "__main__":
    xml_file = "resultXML.xml"  # Substitua pelo nome do seu arquivo
    tags, values = extract_tags(xml_file)
    
    print("Tags únicas encontradas no XML:")
    for tag, value in zip(tags, values):
        print(f"{tag} : {value}")
```
